```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import scipy.stats
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Set visual style
sns.set(style="whitegrid")
```

```python
# Load the Boston housing dataset from the URL provided in the course
URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ST0151EN-SkillsNetwork/labs/boston_housing.csv'
boston_df = pd.read_csv(URL)

# Preview the first few rows
boston_df.head()
```

```python
# 1. Boxplot for "Median value of owner-occupied homes" (MEDV)
plt.figure(figsize=(8, 5))
sns.boxplot(y=boston_df['MEDV'])
plt.title("Boxplot of Median Value of Owner-Occupied Homes (MEDV)")
plt.ylabel("MEDV (in $1000's)")
plt.show()
```

```python
# 2. Histogram for the Charles River variable (CHAS)
plt.figure(figsize=(6, 4))
sns.countplot(x='CHAS', data=boston_df)
plt.title("Number of Homes Bounded by the Charles River")
plt.xlabel("CHAS (1 = Tract bounds river; 0 = Otherwise)")
plt.ylabel("Count")
plt.show()
```

```python
# 3. Boxplot of MEDV vs AGE (Discretized Age Groups)
bins = [-1, 35, 70, float('inf')]
labels = ['35 years and younger', 'Between 35 and 70 years', '70 years and older']
boston_df['AGE_GROUP'] = pd.cut(boston_df['AGE'], bins=bins, labels=labels)

plt.figure(figsize=(10, 6))
sns.boxplot(x='AGE_GROUP', y='MEDV', data=boston_df)
plt.title("MEDV by Age Group")
plt.xlabel("Age Group")
plt.ylabel("MEDV")
plt.show()
```

```python
# 4. Scatter plot: Nitric Oxide Concentrations (NOX) vs Proportion of Non-Retail Business Acres (INDUS)
plt.figure(figsize=(8, 5))
sns.scatterplot(x='INDUS', y='NOX', data=boston_df)
plt.title("Relationship between NOX and INDUS")
plt.xlabel("Proportion of Non-Retail Business Acres (INDUS)")
plt.ylabel("Nitric Oxides Concentration (NOX)")
plt.show()
```

```python
# 5. Histogram for the Pupil-Teacher Ratio (PTRATIO)
plt.figure(figsize=(8, 5))
sns.histplot(boston_df['PTRATIO'], bins=20, kde=True)
plt.title("Histogram of Pupil-Teacher Ratio (PTRATIO)")
plt.xlabel("Pupil-Teacher Ratio")
plt.ylabel("Frequency")
plt.show()
```

```python
# Task 1: T-test for MEDV based on Charles River binding (CHAS)
# H0: No significant difference in median value between homes bounded/not bounded by Charles River
# H1: Significant difference in median value
charles_yes = boston_df[boston_df['CHAS'] == 1]['MEDV']
charles_no = boston_df[boston_df['CHAS'] == 0]['MEDV']

t_stat, p_value = scipy.stats.ttest_ind(charles_yes, charles_no, equal_var=True)
print(f"T-statistic: {t_stat:.4f}, P-value: {p_value:.4e}")

if p_value < 0.05:
    print("Conclusion: Reject the null hypothesis. There is a significant difference.")
else:
    print("Conclusion: Fail to reject the null hypothesis.")
```

```python
# Task 2: ANOVA for MEDV across AGE_GROUP
# H0: No significant difference in median value across age groups
# H1: At least one age group has a different median value
anova_model = ols('MEDV ~ C(AGE_GROUP)', data=boston_df).fit()
anova_table = sm.stats.anova_lm(anova_model, typ=2)
print(anova_table)
```

```python
# Task 3: Pearson Correlation between NOX and DIS
# H0: Nitric oxide concentration is not correlated with weighted distances to employment centers
# H1: Nitric oxide concentration is correlated with weighted distances
corr, p_val = scipy.stats.pearsonr(boston_df['NOX'], boston_df['DIS'])
print(f"Pearson Correlation Coefficient: {corr:.4f}, P-value: {p_val:.4e}")
```
